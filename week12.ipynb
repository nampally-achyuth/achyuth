```python
import matplotlib.pyplot as plt
import pandas as pd

data={
    "month":['jan','feb','mar','apr','may','june'],
    "north":[100, 110, 120, 130, 140, 150],
    "south":[80, 85, 90, 95, 100, 105],
    "east":[120, 125, 130, 135, 140, 145],
    "west":[300, 290, 280, 270, 260, 250]
    }

df = pd.DataFrame(data)
display(df.head())

plt.figure(figsize=(10, 6))
plt.plot(df['month'], df['north'], label='North', marker='o')
plt.plot(df['month'], df['south'], label='South', marker='x')
plt.plot(df['month'], df['east'], label='East', marker='^')
plt.plot(df['month'], df['west'], label='West', marker='s')
plt.title("Overall Monthly Sales by Region")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

#sales comparison across regions
regions=['north','south','east','west']
total=[df['north'].sum(),df['south'].sum(),df['east'].sum(),df['west'].sum()]
plt.bar(regions,total,color=['red','green','blue','yellow']) # Fixed typo: 'colour' to 'color'
plt.title("Sales Comparison Across Regions")
plt.xlabel("Regions")
plt.ylabel("Total Sales")
plt.show()


#catagory -wise sales distribution(piechart)
labels=['north','south','east','west']
sizes=[df['north'].sum(),df['south'].sum(),df['east'].sum(),df['west'].sum()]
plt.pie(sizes,labels=labels,autopct='%1.1f%%')
plt.title("Category-wise Sales Distribution")
plt.show()



#monthly sales distribution across  regions(stacked  bar code)
plt.figure(figsize=(10, 6))
plt.bar(df['month'], df['north'], label='North', color='blue')
plt.bar(df['month'], df['south'], label='South', color='green', bottom=df['north'])
plt.bar(df['month'], df['east'], label='East', color='red', bottom=df['north'] + df['south'])
plt.bar(df['month'], df['west'], label='West', color='yellow', bottom=df['north'] + df['south'] + df['east'])
plt.title("Monthly Sales Distribution Across Regions (Stacked Bar Chart)")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
